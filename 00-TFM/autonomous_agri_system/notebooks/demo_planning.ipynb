# Hybrid Planning Demo
Global + local trajectory selection prototype.

```python
import numpy as np
from planning.global_planner import BoustrophedonPlanner
from planning.local_planner import HybridLocalPlanner, PlannerState
from planning.trajectory_scorer_net import TrajectoryScorerNet

occ = np.zeros((200,200), dtype=np.float32)
occ[90:110, 95:105] = 1
global_plan = BoustrophedonPlanner(row_step=10).plan((1-occ).astype(np.uint8))
local = HybridLocalPlanner(TrajectoryScorerNet())
best, idx, all_traj = local.select(occ, PlannerState(20,20,0.0,1.0), goal=(180,180))
global_plan.waypoints.shape, idx, best.shape, all_traj.shape
```
