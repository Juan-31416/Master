# BEV Perception Demo
Quick walkthrough for synthetic BEV data and model forward pass.

```python
from perception.bev_transformer import BEVConfig, BEVTransformer
from perception.data_generator import SyntheticBEVConfig, SyntheticBEVDataset

ds = SyntheticBEVDataset(SyntheticBEVConfig(num_samples=2))
sample = ds[0]
model = BEVTransformer(BEVConfig())
out = model(sample['camera'].unsqueeze(0), sample['lidar'].unsqueeze(0))
{k: v.shape for k,v in out.items()}
```
